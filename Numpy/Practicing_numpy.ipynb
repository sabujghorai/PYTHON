```python
import numpy as np
a1 = np.array([[1,2,3,4],
               [4,5,6,7,]])
a2 = np.array([1.1,2.2,3.3,4.4])
print(a1.dtype)
print(a2.dtype)

print(a1.itemsize)
print(a2.itemsize)

print(f"The shape is : {a1.shape}")
print("There are ",a1.size,"elements in a1")

print(f"Dimenstion : {a1.ndim}")
```

```python
# A 1D array containing 10 zeros.
import numpy as np
arr = np.zeros((1,10))
print(arr)
```

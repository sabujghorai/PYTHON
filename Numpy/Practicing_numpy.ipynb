```python
import numpy as np
a1 = np.array([[1,2,3,4],
               [4,5,6,7,]])
a2 = np.array([1.1,2.2,3.3,4.4])
print(a1.dtype)
print(a2.dtype)

print(a1.itemsize)
print(a2.itemsize)

print(f"The shape is : {a1.shape}")
print("There are ",a1.size,"elements in a1")

print(f"Dimenstion : {a1.ndim}")
```

```python
# A 1D array containing 10 zeros.
import numpy as np
arr = np.zeros((1,10))
print(arr)
```

```python
# A 3 × 3 matrix containing ones.
arr1 = np.ones((3,3))
print(arr1)
```

```python
# Create a NumPy array containing numbers from 1 to 50.
arra = np.arange(0,51)
print(arra)
```

```python
# Create an array containing numbers from 1 to 12

array = np.arange(1,13)
print(array)
print("Reshaped array : \n",array.reshape(3,4))
print("Reshaped again : \n",array.reshape(2,6))
```

```python
arr = np.array([10, 20, 30, 40, 50, 60])
print("First element is :",arr[0])
print("Last element is :",arr[-1])
print("Last element is :",arr[-1:])
print("Elements from 1 to 4 :",arr[1:5])
print("Reversed array is :",arr[::-1])
```

```python
# Given an 2D array find 50,70,30 ,(First row),(last Row),(second row)
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(arr[0,2])
print(arr[1,1])
print(arr[2,0])

print("First Row is :",arr[0])
print("Second Row is :",arr[1])
print("Second Row is :",arr[-1])
```

```python
import numpy as np

arr = np.array([2, 4, 6, 8, 10])
for i in range(len(arr)):
    print(arr[i] + 5)
```

```python
import numpy as np
marks = np.array([78, 85, 92, 67, 88, 95, 72])
print(marks.sum(axis=0))
print(marks.mean(axis=0))
print(marks.max(axis=0))
print(marks.min(axis=0))
print(marks.std())
```

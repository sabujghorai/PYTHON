## Minumym sales per restaurent

```python
import numpy as np
sales_data =np.array([
# [Restaurent :Id , 2021 ,  2022 ,  2023 ,  2024]
            [  1, 150000, 180000, 220000, 250000], # Paradise Biriyani
            [  2, 120000, 140000, 160000, 190000], # Pizza Hub
            [  3, 200000, 230000, 260000, 300000], # Burger Point 
            [  4, 180000, 210000, 240000, 270000], # Chai house
            [  5, 160000, 185000, 200000, 230000], # Coffe House

])

yearly_total = sales_data[:, 1:].sum(axis=0)
print(f"Total income in a year of each shop is is : {yearly_total}")

# minimum sales per restaurent
min_sales = sales_data[:, 1:].min(axis=1) # axis 1 means in a row which is the minimun earning
print(min_sales)

# maximum sales per restaurent
maximum_sales = sales_data[:, 1:].max(axis=0) # axis 0 means in a column which is the max earning
print(f"Maximum earning of the restaurent : {maximum_sales}")

# Average sales per restauresnt
avg_sales = sales_data[:, 1:].mean(axis=0) # average of the column
print(f"Average earning per year : {avg_sales}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
sales_data =np.array([
# [Restaurent :Id , 2021 ,  2022 ,  2023 ,  2024]
            [  1, 150000, 180000, 220000, 250000], # Paradise Biriyani
            [  2, 120000, 140000, 160000, 190000], # Pizza Hub
            [  3, 200000, 230000, 260000, 300000], # Burger Point 
            [  4, 180000, 210000, 240000, 270000], # Chai house
            [  5, 160000, 185000, 200000, 230000], # Coffe House

])

cumilative_sum = np.cumsum(sales_data[:, 1:],axis=1)
print(cumilative_sum)
plt.figure(figsize=(10,6))
plt.plot(np.mean(cumilative_sum,axis=0))
plt.title("Average cumelative sales per year ")
plt.xlabel("Years ---->")
plt.ylabel("Sales ---->")
plt.grid(True)
plt.show()
```
